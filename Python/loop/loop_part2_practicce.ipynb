check wheather number contains digit 7

```python
num = 145738 #yes/No
```

```python
found = False
while num>0:
  digit = num %10 #8 3 7
  if digit ==7:
    found = True
    break

  num = num//10
  print(num)

if found:
  print("yes")
else:
  print("No")
```

check wheather a number is palindrome # 121, 123

```python
reverse = 0
num = 1221
original = num
while num >0:
  digit = num %10  #5 3
  reverse = reverse *10 + digit # 5 3
  num = num//10 #123

print(reverse)
# print(num)
if reverse == original:
  print("yes")
else:
  print("No")
```
