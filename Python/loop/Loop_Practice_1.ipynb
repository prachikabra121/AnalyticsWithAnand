find factorial of a number 5! = 5x4x3x2x1

```python
num = 5
fact = 1
for i in range(1,num+1):
  fact = fact *i
print(fact)
```

find the maximum number from the list without using max function

```python
l = [10,21,0,4,101,899987,56,34]
```

```python
largest = l[0]

for i in l:
  if i>largest:
    largest = i
print(largest)
```

pyramid

```python
rows = 5
for i in range(rows):
  print(" " * (rows-i-1) + "*" * (2*i + 1))
```
