Question 1:

Hint: Your output should display each item on a new line.

```python

for i in l :
  # print(i)
  if type(i) == dict:
    for k,v in i.items():
      print(k,":",v)
      if type(v)== set:
        for s in v:
          print(s)

  elif type(i)==tuple:
    for t in i:
      print(t)

  elif type(i)==set:
    for r in i:
      print(r)

  elif type(i)== list:
    for h in i:
      print(h)
  else:
    print(i)
```

```python
l = [
    10,
    "Python",
    [20, 30, [40, 50]],
    (60, 70),
    {
        "name": "Vaibhav",
        "age": 25,
        "skills": ["SQL", "AWS"]
    }
]


for item in l:

    if type(item) == list:
        print("List:", item)

        for sub_item in item:
            print("  ", sub_item)

    elif type(item) == tuple:
        print("Tuple:", item)

        for sub_item in item:
            print("  ", sub_item)

    elif type(item) == dict:
        print("Dictionary:", item)

        for key, value in item.items():
            print("  Key:", key)
            print("  Value:", value)

    elif type(item) == int:
        print("Integer:", item)

    elif type(item) == str:
        print("String:", item)

    else:
        print("Other:", item)
```

Question 2:  
Identify the type of each element in the list.  
Hint: Use a for loop to iterate through the list and print the type of each element.  

```python
for i in l:
  print(f"{i} --->{type(i).__name__}")
```

```python
l1=[]
for i in l :
  if type(i)==int:
    l1.append(i)
  elif type(i) == dict:
    for k,v in i.items():
      print(k,":",v)
      if type(v) == int:
        l1.append(v)
      elif type(v)== set:
        for s in v:
          if type(s)==int:
            l1.append(s)

  elif type(i)==tuple:
    for t in i:
      if type(t)==int:
        l1.append(t)

  elif type(i)==set:
    for r in i:
      if type(r)==int:
        l1.append(r)

  elif type(i)== list:
    for h in i:
      if type(h)==int:
        l1.append(h)
  else:
    print(i)
```
