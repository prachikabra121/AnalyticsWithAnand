```python
balance = 10000
card = "1234567890"
correct_pin = 1234

def check_balnce():
  print(f"Your balance is {balance}")

def verify_card():
  enter_card = input("Please swipe /insert your card:")
  if enter_card == card:
    return True

  else:
    print("invalid card...")
    return False
```

```python
def verify_pin():
  pin =int(input("enter your pin...."))
  if pin == correct_pin:
    return True
  else:
    print("Incorrect pin")
    return False
```

```python
verify_card()
```

```python
verify_pin()
```

```python
if verify_card():
  if verify_pin():
    print("\n Login successfull!")
    print("Welcome to ATM")
    check_balnce()
  else:
    print("Transaction failed")

else:
  print("Please insert valid card")
```

```python
def withdraw_balance():
  amount = int(input("enter the amount"))
  print(f"withdraw amount successful{amount}")
  balance1 = balance- amount
  print(f"your balance is {balance1}")
```

```python
def withdraw_balnace(amount):
  global balance
  if amount > balance:
    print("Insufficient balnace")
  else:
    balance = balance-amount
    print(f"Please collect your cash {amount}")
    print(f"Remaining balance: Rs{balance}")
```

```python
if verify_card():
  if verify_pin():
    print("\n Login successfull!")
    print("Welcome to ATM")
    amount = int(input("enter withdrwal amoyunt: "))
    withdraw_balnace(amount)

  else:
    print("Transaction failed")

else:
  print("Please insert valid card")
```
