```python
def login_required(func):
  def wrapper(username):
    if username == "admin":
      func(username)
    else:
      print("Access Denied")
  return wrapper
```

```python
@login_required
def delete_data(username):
  print("Data deleted")
```

```python
users = {
    "Anand": "admin",
    "Rahul": "user",
    "Priya": "manager"
}
```

```python
def admin_only(func):
  def wrapper(username):
    if users.get(username) == "admin":
      func(username)
    else:
      print("Access Denied")
  return wrapper
```

```python
@admin_only
def delete_data(username):
  print("Data deleted")
```

```python
delete_data("Anand")
```

```python
delete_data("Priya")
```

```python
users = {
    "Anand": ["read", "write", "delete"],
    "Rahul": ["read"],
    "Priya": ["read", "write"]
}


def requires_permission(permission):

    def decorator(func):

        def wrapper(username):

            if permission in users.get(username, []):
                func(username)
            else:
                print("Access Denied")

        return wrapper

    return decorator
```

```python
@requires_permission("delete")
def delete_data(username):
    print(username, "deleted the data")


@requires_permission("write")
def update_data(username):
    print(username, "updated the data")
```

```python
delete_data("Priya")
```

```python
delete_data("Anand")
```

```python
update_data("Priya")
```

```python
update_data("Rahul")
```
