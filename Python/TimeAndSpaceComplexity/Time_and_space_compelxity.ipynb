<h1>Why Do We Care About Complexity?</h1>

<h3>When programs deal with large data, two things matter:</h3>

<b>How fast does code run? → Time Complexity<b>

<b>How much memory does code use? → Space Complexity<b>

<h4>For small data, everything looks fast.
For millions of rows, bad code becomes slow or crashes.
</h4>

Example:

Searching 10 records → instant

Searching 10 million records → can be slow if poorly designed

<h2>What is Time Complexity?</h2>

<b>Time complexity measures:</b>

How execution time grows as input size grows.

We don’t measure seconds — we measure growth behavior.

We use Big-O notation.

<h2>What is Big-O Notation?</h2>

<b>Big-O notation is a mathematical way to describe:

1)How the running time or memory usage of an algorithm grows when input size increases.

2)Instead of measuring time in seconds, we measure how performance scales as data grows.

<h2>Why don’t we measure in seconds?</h2>

<b>Because:

1)Different computers have different speeds.

2)Same code runs differently on different machines.

3)Network, CPU, memory all vary.

So instead, we measure growth pattern.

<h1>How to Calculate Complexity Step-by-Step</h1>

```python
#Step 1: Count loops
for x in arr:
    print(x)  # nX 30 o(n)
```

Loop runs n times.

Complexity: O(n)

```python
#Step 2: Nested loops multiply
for i in arr:
    for j in arr:
        print(i, j)
```

Outer loop: n

Inner loop: n

Total = n × n

Complexity: O(n²)

```python
#Step 3: Consecutive loops add

for x in arr:
    print(x)

for y in arr:
    print(y)
```

<b>Interview : Nested Loop Confusion

Not O(n²), but O(n).

Loops are separate.

<b>Interview : Dictionary Lookup Assumption

x in dict

Average: O(1)

Worst case: O(n)

Total work:

n + n = 2n
Drop constants:

complexity = O(n)



<b>Interview : Recursion Stack Space

```python
def f(n):
    if n == 0:
        return
    f(n-1)
```

Time: O(n)

Space: O(n) due to call stack.

```python
#Step 4: Ignore constants

for x in arr:
    print(x)
    print(x)
```

still :
complexity = O(n)


```python
#Step 5: Drop smaller terms

arr = [1, 2, 3, 4]

# Part 1 → O(n²)
for i in arr:
    for j in arr:
        print(i, j)

# Part 2 → O(n)
for x in arr:
    print(x)
```

Assume array size = n.

Part 1: Nested loops

n × n = n²

Time complexity: O(n²)


Part 2: Single loop

Time complexity: O(n)

Total Complexity

Total work:n² + n
So complexity: O(n² + n)

Big-O Simplification

In Big-O, we keep the dominant term:

so time complexity will be o(n²)

<h3>What is Caching? </h3>

Caching means:

Storing previously computed or frequently used results so we don’t recompute them again.

Compute once → Store → Reuse later

Imagine:

You ask Google the same question multiple times.

Google does not recompute everything every time.
It stores popular results in cache.

So results appear instantly.


```python
#without caching
def square(n):
    print("Computing...")
    return n * n

print(square(4))
print(square(4))
#It recomputes each time.
```

```python
cache = {}

def square(n):
    if n not in cache:
        print("Computing...")
        cache[n] = n * n
    return cache[n]

print(square(4))
print(square(4))
```

```python
print(square(4))
```

```python
print(square(10))
print(square(5))
print(square(8))
```
