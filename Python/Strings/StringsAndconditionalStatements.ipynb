```python
salary = int(input("Enter your Salary: "))

if salary >= 50000:

    score = int(input("Enter your Credit Score: "))

    if score >= 750:
        print("Loan Approved")

    elif score >= 650:
        print("Manual Verification Required")

        document = input("Are your documents verified? (yes/no): ")

        if document.lower() == "yes":
            print("Documents Verified")
            print("Loan Approved")
        else:
            print("Documents Not Verified")
            print("Loan Rejected")

    else:
        print("Credit Score is Low")
        print("Loan Rejected")

else:
    print("Salary is below eligibility criteria")
```
